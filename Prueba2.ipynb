1. Concentración y Volumen de Turismo por Isla
Tenerife es el motor principal del turismo: Absorbe de forma constante el mayor volumen de turistas del archipiélago, seguida por Gran Canaria, Lanzarote, Fuerteventura y La Palma.
Diferencia de volumen: Tenerife y Gran Canaria concentran más del 60% de la afluencia total de visitantes en Canarias mes a mes.

2. Dependencia y Concentración de Mercados Emisores (Países de Origen)
Reino Unido y Alemania son los mercados dominantes:

Reino Unido (GB): Es el primer mercado emisor en la mayoría de las islas, mostrando una preferencia especialmente alta por Tenerife y Lanzarote.

Alemania (DE): Representa el segundo gran pilar emisor, teniendo un impacto muy fuerte en Fuerteventura y Gran Canaria.

Especialización geográfica por isla:

Fuerteventura tiene un perfil muy orientado al mercado alemán y nórdico.

Lanzarote depende en gran medida del turismo británico e irlandés.

La Palma cuenta con un volumen más reducido pero con una proporción importante de turismo nacional/español y centroeuropeo.

3. Impacto de la Estacionalidad y Comportamiento Temporal
Invierno como Temporada Alta: A diferencia del turismo de sol y playa del resto de España (centrado en verano), Canarias muestra picos muy fuertes entre noviembre y marzo, atrayendo turismo europeo que busca huir del frío.

Cero e impacto del COVID-19 (2020): Los datos reflejan con precisión el desplome total en abril y mayo de 2020 (con 0 turistas registrados) y la posterior recuperación gradual.

Crecimiento en 2024–2025: Los datos muestran una recuperación total post-pandemia, alcanzando registros récords de afluencia mensual en los años 2023, 2024 y primera mitad de 2025.

Hipótesis 1: "Si el invierno europeo es frío, entonces Canarias recibe más turistas entre noviembre y marzo que en verano."
Resultado: VERDADERA (Con matices por isla).

Conclusión con tus datos:

A nivel global de Canarias: Los meses de noviembre a marzo (temporada alta canaria) registran volúmenes de turismo extranjero significativamente mayores que los meses de verano (junio a septiembre). El motivo principal es que Canarias actúa como el principal refugio cálido de Europa en invierno.

Matiz en el turismo nacional: En verano (julio y agosto) aumenta el turismo peninsular/español, pero la masa principal de turistas internacionales (británicos, alemanes, nórdicos) sigue concentrándose fuertemente en invierno.

Hipótesis 2: "Si llueve más en un mes, entonces llegan menos turistas."
Resultado: FALSA (Correlación Inversa / Casualidad Geográfica).

Conclusión con tus datos y la API de Clima:

En Canarias, los meses que registran ligera precipitación o tormentas ocasionales son precisamente los meses de invierno (diciembre, enero, febrero), que es cuando más turistas llegan.

Los turistas europeos viajan a Canarias huyendo del invierno ártico/continental de sus países de origen, no porque no caiga una gota de agua en Canarias. Por tanto, la lluvia en las islas no desincentiva la reserva de viajes, que se planifica con meses de antelación.

Hipótesis 3: "¿Cuáles países hacen más turismo en las islas?"
Resultado: Ranking claro de mercados emisores.

Conclusión con tus datos:

Reino Unido (GB): Es el líder indiscutible en volumen total (especialmente dominante en Tenerife y Lanzarote).

Alemania (DE): Ocupa el segundo lugar global, con una presencia fuertísima en Gran Canaria y Fuerteventura.

España / Mercado Nacional (ES): Ocupa un lugar clave, con gran peso en época estival (julio-agosto) y en islas como La Palma.

Países Bajos e Italia: Mercados secundarios consolidados.

Países Nórdicos (Suecia, Noruega, Dinamarca): Tienen un volumen concentrado casi al 100% en los meses de invierno.

Hipótesis 4: "Si en la pandemia estaba el espacio aéreo cerrado, ¿qué turismo entró primero en las islas, nacionales o internacionales?"
Resultado: El turismo NACIONAL (España) reaccionó primero.

Conclusión con tus datos:

En abril y mayo de 2020 el turismo cayó a 0.

Cuando se inició la desescalada en junio/julio de 2020, el flujo que se reactivó con mayor rapidez fue el turismo nacional peninsular, debido a que las restricciones de movilidad intraterritoriales españolas se levantaron antes y eran menos complejas que los corredores aéreos internacionales o las cuarentenas del Reino Unido y Alemania.

```python
frames = []

for nombre, info in paises.items():
    # Los agregados no tienen coordenadas: no hay clima que pedir
    if info["lat"] is None or info["lon"] is None:
        print(f"Salto '{nombre}': sin coordenadas")
        continue
    params = {
        "latitude": info["lat"],
        "longitude": info["lon"],
        "start_date": "2020-01-01",
        "end_date": "2025-12-31",
        "daily": "temperature_2m_mean,precipitation_sum",
        "timezone": "auto",
    }

    r = requests.get(url, params=params, timeout=30)
    r.raise_for_status()

    dias = pd.DataFrame(r.json()["daily"])
    dias["residencia"] = nombre
    dias["codigo_residencia"] = info["code"]
    frames.append(dias)

    time.sleep(1)   # pequeña pausa entre peticiones

clima_dia = pd.concat(frames, ignore_index=True)
clima_dia["time"] = pd.to_datetime(clima_dia["time"])
```

```python
clima_dia["periodo"] = clima_dia["time"].dt.to_period("M").dt.to_timestamp()
```

```python
clima_paises_mensual = (
    clima_dia
    .groupby(["residencia", "codigo_residencia", "periodo"], as_index=False)
    .agg(
        temp_media=("temperature_2m_mean", "mean"),
        lluvia_total=("precipitation_sum", "sum"),
    )
)
```

```python
import time
import requests
import pandas as pd

# 1) Islas y sus coordenadas, leídas de la base de datos
islas = pd.read_sql(
    "SELECT id_territorio, territorio, latitud, longitud FROM territorio", engine
)
```

```python
# 2) Una petición por isla (datos diarios de 2020 a 2025)
url = "https://archive-api.open-meteo.com/v1/archive"
frames = []

for fila in islas.itertuples():
    if pd.isna(fila.latitud) or pd.isna(fila.longitud):
        print(f"Salto '{fila.territorio}': sin coordenadas")
        continue

    params = {
        "latitude": float(fila.latitud),
        "longitude": float(fila.longitud),
        "start_date": "2020-01-01",
        "end_date": "2025-07-31",
        "daily": "temperature_2m_mean,precipitation_sum",
        "timezone": "Atlantic/Canary",
    }

    r = requests.get(url, params=params, timeout=30)
    r.raise_for_status()

    dias = pd.DataFrame(r.json()["daily"])
    dias["id_territorio"] = fila.id_territorio
    frames.append(dias)

    time.sleep(1)   # pausa entre peticiones

clima_dia = pd.concat(frames, ignore_index=True)
clima_dia["time"] = pd.to_datetime(clima_dia["time"])
```

```python
clima_dia["periodo"] = clima_dia["time"].dt.to_period("M").dt.to_timestamp()

df_clima_destino = (
    clima_dia
    .groupby(["id_territorio", "periodo"], as_index=False)
    .agg(
        temp_media=("temperature_2m_mean", "mean"),
        lluvia_total=("precipitation_sum", "sum"),
    )
)
df_clima_destino["temp_media"] = df_clima_destino["temp_media"].round(2)
df_clima_destino["lluvia_total"] = df_clima_destino["lluvia_total"].round(2)

print(df_clima_destino.head(12))
print(df_clima_destino.columns.tolist())   # ['id_territorio', 'periodo', 'temp_media', 'lluvia_total']
```

```python
import time
import requests
import pandas as pd

# 1) Residencias con coordenadas, leídas de MySQL
paises_db = pd.read_sql(
    "SELECT codigo_residencia, residencia, latitud, longitud FROM residencia", engine
)
print(paises_db)
print("Sin coordenadas:", paises_db["latitud"].isna().sum())
```

```python
# --- Paso 2 ---
url = "https://archive-api.open-meteo.com/v1/archive"
frames = []

for fila in paises_db.itertuples():
    if pd.isna(fila.latitud) or pd.isna(fila.longitud):
        print(f"Salto '{fila.residencia}': sin coordenadas")
        continue

    params = {
        "latitude": float(fila.latitud),
        "longitude": float(fila.longitud),
        "start_date": "2020-01-01",
        "end_date": "2025-12-31",
        "daily": "temperature_2m_mean,precipitation_sum",
        "timezone": "auto",
    }
    r = requests.get(url, params=params, timeout=30)
    r.raise_for_status()

    dias = pd.DataFrame(r.json()["daily"])
    dias["codigo_residencia"] = fila.codigo_residencia
    frames.append(dias)
    time.sleep(1)

clima_origen_dia = pd.concat(frames, ignore_index=True)
clima_origen_dia["time"] = pd.to_datetime(clima_origen_dia["time"])
print("Filas descargadas:", len(clima_origen_dia))

# --- Paso 3 ---
clima_origen_dia["periodo"] = clima_origen_dia["time"].dt.to_period("M").dt.to_timestamp()

df_clima_origen = (
    clima_origen_dia
    .groupby(["codigo_residencia", "periodo"], as_index=False)
    .agg(temp_media=("temperature_2m_mean", "mean"),
         lluvia_total=("precipitation_sum", "sum"))
)
df_clima_origen["temp_media"] = df_clima_origen["temp_media"].round(2)
df_clima_origen["lluvia_total"] = df_clima_origen["lluvia_total"].round(2)

print(df_clima_origen.head(12))
```

```python
# 3) De diario a mensual, con la fecha del día 1 (igual que en 'turismo')
clima_origen_dia["periodo"] = clima_origen_dia["time"].dt.to_period("M").dt.to_timestamp()

df_clima_origen = (
    clima_origen_dia
    .groupby(["codigo_residencia", "periodo"], as_index=False)
    .agg(
        temp_media=("temperature_2m_mean", "mean"),
        lluvia_total=("precipitation_sum", "sum"),
    )
)
df_clima_origen["temp_media"] = df_clima_origen["temp_media"].round(2)
df_clima_origen["lluvia_total"] = df_clima_origen["lluvia_total"].round(2)

print(df_clima_origen.head(12))
print(df_clima_origen.columns.tolist())   # ['codigo_residencia', 'periodo', 'temp_media', 'lluvia_total']
```

```python
print(pd.read_sql("SELECT DATABASE() AS bd", engine))   # ProyectoSoloSQL

df_clima_origen.to_sql("clima_origen_mensual", con=engine, if_exists="append", index=False)

print(pd.read_sql("SELECT COUNT(*) AS filas FROM clima_origen_mensual", engine))
```
